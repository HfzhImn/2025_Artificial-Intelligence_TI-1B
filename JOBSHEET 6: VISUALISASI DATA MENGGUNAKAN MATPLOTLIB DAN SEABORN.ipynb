## Melakukan Visualiasi Data menggunakan Pie Chart

```python
import matplotlib.pyplot as plt

flavors = ('Chocolate', 'Vanilla', 'Pistachio', 'Mango', 'Strawberry')
votes = (12, 11, 4, 8, 7)

colors = ('#8B4513', '#FFF8DC', '#93C572', '#E67F0D','#D53032')

explode = (0, 0, 0, 0.1, 0)

plt.pie(votes,
        labels=flavors,
       autopct='%1.2f%%',
       colors=colors,
       explode = explode,
       shadow = True)
plt.show()
```

## Melakukan Visualisasi Data menggunakan Bar Chart

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

countries = ('Argentina', 'Bolivia', 'Brazil', 'Chile',  'Colombia', 'Ecuador', 'Falkland Islands', 'French Guiana', 'Guyana', 'Paraguay', 'Peru','Suriname', 'Uruguay', 'Venezuela')

populations = (45076704, 11626410, 212162757, 19109629,50819826, 17579085, 3481, 287750, 785409, 7107305, 32880332,585169, 3470475, 28258770)

df = pd.DataFrame({
    'Country' : countries,
    'Population' : populations,
})

df.sort_values(by='Population', inplace = True)

x_coords = np.arange(len(countries))

plt.bar(x_coords, df['Population'], tick_label=df['Country'])
plt.xticks(rotation=90) #rotates text for x-axis labels
plt.ylabel('Population')
plt.title('South American Populations')
plt.show()


```

## Latihan 2: Permen
Mario memiliki sekantong permen. Terdapat lima jenis permen, masing-masing diberi nama di bawah ini. Buat diagram yang menunjukkan persentase peluang bahwa kita akan mengeluarkan permen Snickers dari kantong jika mario melakukan pengambilan acak. Sebutkan peluang memilih permen Snickers.


```python
import matplotlib.pyplot as plt

# Data
candy_names = ['Kit Kat', 'Snickers', 'Milky Way', 'Toblerone', 'Twix']
candy_counts = [52, 39, 78, 13, 78]

# Hitung persentase
total_candies = sum(candy_counts)
snickers_percentage = (candy_counts[1] / total_candies) * 100

# Pie chart
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']
plt.figure(figsize=(7, 7))
plt.pie(candy_counts, labels=candy_names, autopct='%1.1f%%', startangle=140, colors=colors)
plt.title('Distribusi Permen dalam Kantong Mario')
plt.axis('equal')  # Membuat pie chart berbentuk lingkaran
plt.show()

# Cetak peluang Snickers
print(f"Peluang memilih permen Snickers: {snickers_percentage:.1f}%")
```

## Latihan 3: Makanan
Restoran memiliki menu makanan penutup yang terlalu besar. Mereka ingin memotong beberapa item dari menu. Untuk membuat sebagian besar pelanggan mereka senang, mereka ingin menghapus hanya tiga makanan penutup yang paling tidak populer dari menu. Kita memiliki daftar makanan penutup yang disajikan restoran, serta hitungan berapa kali makanan penutup tersebut dijual dalam seminggu terakhir. Buat visualisasi yang menunjukkan popularitas relatif dari makanan penutup. Sebutkan tiga makanan penutup yang harus disingkirkan.


```python
import matplotlib.pyplot as plt

# Data penjualan
dessert_sales = {
    'Lava Cake': 14,
    'Mousse': 5,
    'Chocolate Cake': 12,
    'Ice Cream': 19,
    'Truffles': 6,
    'Brownie': 8,
    'Chocolate Chip Cookie': 12,
    'Chocolate Pudding': 9,
    'Souffle': 10,
    'Chocolate Cheesecake': 17,
    'Chocolate Chips': 2,
    'Fudge': 9,
    'Mochi': 13,
}

# Urutkan data dari yang paling sedikit ke paling banyak
sorted_sales = dict(sorted(dessert_sales.items(), key=lambda x: x[1]))

# Plot horizontal bar chart
plt.figure(figsize=(10, 7))
plt.barh(list(sorted_sales.keys()), list(sorted_sales.values()), color='chocolate')
plt.xlabel('Jumlah Terjual')
plt.title('Popularitas Makanan Penutup dalam Seminggu')
plt.grid(axis='x')
plt.tight_layout()
plt.show()

# Tiga makanan dengan penjualan terendah
least_sold = list(sorted_sales.items())[:3]
print("Tiga makanan penutup yang perlu dihapus dari menu:")
for dessert, sales in least_sold:
    print(f"- {dessert} ({sales} kali terjual)")
```

Kita memiliki penggunaan CPU rata-rata per jam untuk komputer pekerja selama seminggu. Setiap baris data mewakili satu hari dalam seminggu yang dimulai dengan Senin. Setiap kolom data adalah satu jam dalam sehari dimulai dengan 0 menjadi tengah malam.

•	Jam berapa pekerja biasanya makan siang?
