## Soal-1
Buatlah table list Top 5 Customers yang memiliki nilai transaksi paling banyak
---





```python
import pandas as pd
import datetime as dt

# Load semua dataset
df_customer = pd.read_csv('sample data/customers.csv')
df_gasstation = pd.read_csv('sample data/gasstations.csv')
df_product = pd.read_csv('sample data/products.csv')
df_transaction = pd.read_csv('sample data/transactions_1k.csv')
df_yearmonth = pd.read_csv('sample data/yearmonth.csv')

# Merge semua tabel
df_merged = df_transaction.merge(
    df_customer, on='CustomerID', how='left'
).merge(
    df_gasstation, on='GasStationID', how='left'
).merge(
    df_product, on='ProductID', how='left'
)

# Konversi tipe data waktu
df_merged['Date'] = pd.to_datetime(df_merged['Date'])
df_merged['Hour'] = pd.to_datetime(df_merged['Time']).dt.hour

# Group by CustomerID dan hitung total Amount dan Price
top_customers = df_merged.groupby('CustomerID').agg({
    'Amount': 'sum',
    'Price': 'sum'
}).sort_values('Price', ascending=False).head(5)

# Rename kolom Price menjadi Total Sales
top_customers = top_customers.rename(columns={'Price': 'Total Sales'})

# Tampilkan hasil
top_customers
```

## Soal-4
Buatlah deskripsi statistik untuk masing masing hari (23,24,25,26) pada dataset transaction_1k
---

```python
import pandas as pd
import datetime as dt

# Load semua dataset
df_customer = pd.read_csv('sample data/customers.csv')
df_gasstation = pd.read_csv('sample data/gasstations.csv')
df_product = pd.read_csv('sample data/products.csv')
df_transaction = pd.read_csv('sample data/transactions_1k.csv')
df_yearmonth = pd.read_csv('sample data/yearmonth.csv')

# Merge semua tabel
df_merged = df_transaction.merge(
    df_customer, on='CustomerID', how='left'
).merge(
    df_gasstation, on='GasStationID', how='left'
).merge(
    df_product, on='ProductID', how='left'
)

# 1. Pastikan kolom Date sudah datetime
df_merged['Date'] = pd.to_datetime(df_merged['Date'])

# 2. Filter tanggal 23-26 Agustus 2012
df_filtered = df_merged[
    df_merged['Date'].isin(pd.to_datetime(['2012-08-23', '2012-08-24', '2012-08-25', '2012-08-26']))
]

# 3. Hitung statistik dengan sintaks yang benar
daily_stats = df_filtered.groupby('Date').agg(
    Amount_count=('Amount', 'count'),
    Amount_mean=('Amount', 'mean'),
    Amount_std=('Amount', 'std'),
    Amount_min=('Amount', 'min'),
    Amount_q1=('Amount', lambda x: x.quantile(0.25)),
    Amount_median=('Amount', 'median'),
    Amount_q3=('Amount', lambda x: x.quantile(0.75)),
    Amount_max=('Amount', 'max'),
    Sales_count=('Price', 'count'),
    Sales_mean=('Price', 'mean'),
    Sales_std=('Price', 'std'),
    Sales_min=('Price', 'min'),
    Sales_q1=('Price', lambda x: x.quantile(0.25)),
    Sales_median=('Price', 'median'),
    Sales_q3=('Price', lambda x: x.quantile(0.75)),
    Sales_max=('Price', 'max')
)

# Tampilkan hasil
daily_stats
```

## Soal-5
Carilah waktu terbaik (hari dan jam) dimana paling banyak user gas station melakukan transaksi
---

```python
import pandas as pd
import datetime as dt

# Load semua dataset
df_customer = pd.read_csv('sample data/customers.csv')
df_gasstation = pd.read_csv('sample data/gasstations.csv')
df_product = pd.read_csv('sample data/products.csv')
df_transaction = pd.read_csv('sample data/transactions_1k.csv')
df_yearmonth = pd.read_csv('sample data/yearmonth.csv')

# Merge semua tabel
df_merged = df_transaction.merge(
    df_customer, on='CustomerID', how='left'
).merge(
    df_gasstation, on='GasStationID', how='left'
).merge(
    df_product, on='ProductID', how='left'
)

# Konversi tipe data waktu
df_merged['Date'] = pd.to_datetime(df_merged['Date'])
df_merged['Hour'] = pd.to_datetime(df_merged['Time']).dt.hour

# Ekstrak jam dari kolom Time
df_merged['Hour'] = pd.to_datetime(df_merged['Time']).dt.hour

# Group by Date dan Hour, lalu hitung total Amount dan Price
best_time = df_merged.groupby(['Date', 'Hour']).agg({
    'Amount': 'sum',
    'Price': 'sum'
}).sort_values('Amount', ascending=False).head(5)

# Rename kolom Price menjadi Gas Sales
best_time = best_time.rename(columns={'Price': 'Gas Sales'})

# Tampilkan hasil
best_time
```
