### Percobaan 3: Implementasi Naïve Bayes

```python
# import library yang digunakan
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load dataset
dataset = pd.read_csv('iklan_sosmed (1).csv', delimiter=';')
dataset.head()
```

```python
X = dataset.iloc[:,2:-1]
y = dataset.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)
```

```python
X_train = X_train.values
y_train = y_train.values
X_test = X_test.values
y_test = y_test.values
```

```python
from sklearn.preprocessing import MinMaxScaler
sc = MinMaxScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
classifier.fit(X_train, y_train)
```

```python
# prediksi X_test
y_pred = classifier.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
# buat confussion matrix & skor akurasi
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Akurasi : {}%".format(acc*100))
print("Confussion Matrix : ")
print(cm)
```

```python
# visualisasi sampel data train
from matplotlib.colors import ListedColormap
X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Klasifier (Sampel Training)')
plt.xlabel('Umur')
plt.ylabel('Perkiraan Gaji')
plt.legend()
```

## Tugas Praktikum

### Nomor 1

####🧪 Percobaan 1 – SVM (Klasifikasi)
🎯 Tujuan: Meningkatkan akurasi model klasifikasi pada dataset Diabetes

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd

# 1. Load data
df = pd.read_csv('diabetes.csv')

# 2. Pisahkan fitur dan label
X = df[df.columns[:8]]
y = df['Outcome']

# 3. Normalisasi
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Split data
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.33, random_state=42)

# 5. Grid Search untuk SVM
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.001, 0.01, 0.1, 1],
    'kernel': ['rbf', 'linear', 'poly']
}

grid = GridSearchCV(SVC(), param_grid, refit=True, cv=5)
grid.fit(X_train, y_train)

# 6. Hasil
print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)

# 7. Uji akurasi di data testing
print("Test Accuracy:", grid.score(X_test, y_test))
```

####🧪 Percobaan 2 – SVR (Regresi)
🎯 Tujuan: Meningkatkan performa prediksi gaji berdasarkan pengalaman kerja

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt

# 1. Load data
data = pd.read_csv('Salary_Data.csv')

# 2. Pisahkan fitur dan label
X = data['YearsExperience'].to_numpy().reshape(-1, 1)
y = data['Salary'].to_numpy()

# 3. Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Grid Search untuk SVR
param_grid = {
    'C': [10, 100, 1000],
    'gamma': [0.01, 0.05, 0.1],
    'kernel': ['rbf', 'linear', 'poly']
}

grid = GridSearchCV(SVR(), param_grid, refit=True, cv=5)
grid.fit(X_scaled, y)

# 5. Hasil
print("Best Parameters:", grid.best_params_)
print("Best Score (CV R^2):", grid.best_score_)

# 6. Evaluasi
y_pred = grid.predict(X_scaled)
print("R² score:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))

# 7. Visualisasi
plt.scatter(X, y, color='blue', label='Data Asli')
plt.plot(X, y_pred, color='red', label='SVR (Best Model)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.title('SVR Prediction vs Actual')
plt.show()
```

### Nomor 2

###🧪Percobaan 3 - Naïve Bayes
🎯 Tujuan: Meningkatkan akurasi model (agar lebih akurat dalam memprediksi apakah seseorang akan melakukan transaksi atau tidak dan lebih stabil dan generalisasi baik saat diuji pada data baru (data testing)).



```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd

# 1. Load data
dataset = pd.read_csv('iklan_sosmed (1).csv', delimiter=';')

# 2. Pisahkan fitur dan label
X = dataset.iloc[:, 2:-1].values  # kolom Umur dan Gaji
y = dataset.iloc[:, -1].values    # kolom Transaksi

# 3. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)

# 4. Pipeline + GridSearchCV
pipeline = Pipeline([
    ('scaler', MinMaxScaler()),
    ('nb', GaussianNB())
])

param_grid = {
    'nb__var_smoothing': [1e-9, 1e-8, 1e-7, 1e-6, 1e-5]
}

grid = GridSearchCV(pipeline, param_grid, cv=5)
grid.fit(X_train, y_train)

# 5. Evaluasi
best_model = grid.best_estimator_
y_pred = best_model.predict(X_test)

acc = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("Best var_smoothing:", grid.best_params_)
print("Akurasi Baru: {:.2f}%".format(acc * 100))
print("Confusion Matrix:\n", cm)
```
