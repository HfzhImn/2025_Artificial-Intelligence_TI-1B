##Praktikum 03

```python
import math
# 1. Trigonometri (Sinus, Kosinus, Tangen)
x = math.sin(math.pi / 2)
y = math.cos(0)
z = math.tan(math.pi / 4)
print("Trigonometri:")
print("sin(90°) =", x)
print("cos(0°) =", y)
print("tan(45°) =", z)
print()

# 2. Logaritma (Natural & Basis 10)
a = math.log(2.71828)
b = math.log10(100)
print("Logaritma:")
print("log(e) =", a)
print("log10(100) =", b)
print()

# 3. Konversi Derajat & Jarak
c = math.degrees(math.pi)
d = math.dist([0, 0], [3, 4])
print("Konversi Derajat & Jarak:")
print("math.degrees(π) =", c)
print("Jarak antara (0,0) dan (3,4) =", d)
print()

# 4. Faktorial
e = math.factorial(5)
print("Faktorial:")
print("5! =", e)
print()

# 5. Operasi Matematika Lainnya (fmod & isnan)
f = math.fmod(20, 3)
g = math.isnan(float('nan'))
print("Operasi Lainnya:")
print("math.fmod(20, 3) =", f)
print("Apakah 'nan' adalah NaN? =", g)
print()

# 6. Fungsi Gamma
h = math.gamma(5)
print("Fungsi Gamma:")
print("Gamma(5) =", h)
print()

# 7. Permutasi
i = math.perm(5, 3)
print("Permutasi:")
print("Permutasi 5P3 =", i)
```

##Praktikum 04

```python

import numpy as np
# -------------------------------------------------
# Bagian A: Operasi Matriks (Penjumlahan Matriks)
# -------------------------------------------------

# Membuat Matriks A dan B menggunakan np.array()
A = np.array([[80, 30, 25],
              [50, 20, 10]])

B = np.array([[30, 60, 15],
              [10, 30, 30]])

# Melakukan operasi penjumlahan matriks A + B
hasil_matriks = A + B

# Menampilkan hasil penjumlahan matriks
print("Hasil Penjumlahan Matriks A + B:")
print(hasil_matriks)
print()


# -------------------------------------------------
# Bagian B: Operasi Dot Product (Perkalian Vektor)
# -------------------------------------------------

# Membuat Vektor A dan B menggunakan np.array()
Vektor_A = np.array([10, 20, 15])
Vektor_B = np.array([16, 67, 35])

# Melakukan operasi dot product menggunakan np.dot()
hasil_dot_product = np.dot(Vektor_A, Vektor_B)

# Menampilkan hasil dari operasi dot product
print("Hasil Dot Product dari Vektor A dan B:")
print(hasil_dot_product)
print()


# -------------------------------------------------
# Bagian C: Operasi Cross Product (Perkalian Silang Vektor)
# -------------------------------------------------

# Menghitung cross product dari Vektor_A dan Vektor_B menggunakan np.cross()
hasil_cross_product = np.cross(Vektor_A, Vektor_B)

# Menampilkan hasil dari operasi cross product
print("Hasil Cross Product dari Vektor A dan B:")
print(hasil_cross_product)
```
