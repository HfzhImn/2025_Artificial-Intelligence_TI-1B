## LATIHAN

### Central Tendency




```python
# Mendapatkan Nilai Mean
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.mean(score)
print("Mean:", x)

# Mendapatkan Nilai Median

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.median(score)
print("Median:", x)

# Mendapatkan Nilai Modus/Mode
from scipy import stats

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = stats.mode(score, keepdims=True)
print("Mode:", x)
```

### Spread

```python
# Mendapatkan Nilai Range
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print("Range:", x)
print("----------------------")

# Mendapatkan Nilai Percentile

data = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.percentile(data, 90)

print("Percentile ke-90:", x)
print("----------------------")

# Mendapatkan Nilai Quartile dan Interquartile Range

data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("Interquartile Range (IQR):", IQR)
print("----------------------")

# Mendapatkan Nilai Variance dan Standart Deviation

data= [86, 87, 88, 86, 87, 85, 86]

mean = np.mean(data)
var = np.var(data)
std = np.std(data)

print("Mean:", mean)
print("Variance:", var)
print("Standard Deviation:", std)
```

### Nomor 1
Pelemparan dadu sebanyak 25 kali. Angka yang keluar datanya adalah :
1 2 3 4 5 5 6 2 3 4 5 6 6 4 3 2 1 4 3 5 6 6 5 4 5
Modus dari data diatas adalah …


```python
from collections import Counter

# Data hasil pelemparan dadu
data = [1, 2, 3, 4, 5, 5, 6, 2, 3, 4, 5, 6, 6, 4, 3, 2, 1, 4, 3, 5, 6, 6, 5, 4, 5]

# Menghitung frekuensi kemunculan setiap angka
frekuensi = Counter(data)

# Mencari nilai modus (angka dengan frekuensi tertinggi)
modus = [angka for angka, jumlah in frekuensi.items() if jumlah == max(frekuensi.values())]

# Menampilkan hasil
print("Data:", data)
print("Frekuensi:", dict(frekuensi))
print("Modus:", modus)
```

### Nomor 2

Diketahui sebuah deret data 9, 10, 11, 6, 8, 7, 7, 5, 4, 5. Tentukan persentil ke-25 dan persentil ke-75 !

```python
import numpy as np

# Data deret
data = [9, 10, 11, 6, 8, 7, 7, 5, 4, 5]

# Mengurutkan data
data_urut = sorted(data)

# Menghitung persentil ke-25 dan ke-75
persentil_25 = np.percentile(data_urut, 25)
persentil_75 = np.percentile(data_urut, 75)

# Menampilkan hasil
print("Data asli      :", data)
print("Data diurutkan :", data_urut)
print("Persentil ke-25:", persentil_25)
print("Persentil ke-75:", persentil_75)
```

### Nomor 3

Pada sebuah kelas memiliki nilai hasil ujian yaitu 50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99. Beradasarkan data hasil ujian tersebut, tentukan :

a.	Urutkan data dari yang terkecil sampai data yang terbesar

b.	Identifikasi nilai first quartile (Q1) dan third quartile (Q3), dan IQR

c.	Tentukan nilai upper bound dan lower bound

d.	Tentukan nilai ujian yang menjadi outlier berdasarkan data hasil ujian diatas



```python
import numpy as np

# Data hasil ujian
data = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]

# a. Urutkan data
data_urut = sorted(data)

# b. Hitung Q1, Q3 dan IQR
Q1 = np.percentile(data_urut, 25)
Q3 = np.percentile(data_urut, 75)
IQR = Q3 - Q1

# c. Hitung lower bound dan upper bound
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# d. Identifikasi outlier
outliers = [x for x in data_urut if x < lower_bound or x > upper_bound]

# Tampilkan hasil
print("a. Data terurut dari terkecil ke terbesar:")
print(data_urut)

print("\nb. Q1 (Kuartil 1):", Q1)
print("   Q3 (Kuartil 3):", Q3)
print("   IQR (Interquartile Range):", IQR)

print("\nc. Lower Bound:", lower_bound)
print("   Upper Bound:", upper_bound)

print("\nd. Outlier dalam data nilai ujian:", outliers)
```

### Nomor 4

Diketahui data berat bada mahasiswa informatika polines kelas IK-2E adalah sebagai berikut.

Berat Badan (kg) dan Frekuensi:

41 – 45	(2)
46 – 50	(3)
50 – 55	(6)
56 – 60	(4)
61 – 65	(5)
66 – 70	(2)

Hitunglah standart deviasi dari data berat badan mahasiswa informatika tersebut.


```python
import math

# Data dari tabel
kelas = [(41, 45), (46, 50), (50, 55), (56, 60), (61, 65), (66, 70)]
frekuensi = [2, 3, 6, 4, 5, 2]

# Hitung nilai tengah setiap kelas
nilai_tengah = [(bawah + atas) / 2 for bawah, atas in kelas]

# Hitung total frekuensi
n = sum(frekuensi)

# Hitung rata-rata (mean) tertimbang
total_fx = sum(f * x for f, x in zip(frekuensi, nilai_tengah))
mean = total_fx / n

# Hitung standar deviasi
total_fx2 = sum(f * (x - mean) ** 2 for f, x in zip(frekuensi, nilai_tengah))
standar_deviasi = math.sqrt(total_fx2 / n)

# Tampilkan hasil
print("Nilai tengah tiap kelas:", nilai_tengah)
print("Total frekuensi (n):", n)
print("Rata-rata (mean):", round(mean, 2))
print("Standar Deviasi:", round(standar_deviasi, 2))
```
